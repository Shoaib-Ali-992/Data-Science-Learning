#  np.where
The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

https://numpy.org/doc/stable/reference/generated/numpy.where.html

```python
a
```

```python
# get indexes through condition
np.where(a>50)
```

```python
# replace values through where given condition
np.where(a>50,1,a)
```

```python
# replace even number through where
np.where(a%2==0,0,a)
```

```python
np.where(a==32,2,a)
```

```python
marks = np.random.randint(1,100,10)
grades = np.where(
    marks >= 80, "A",
    np.where(marks >= 60, "B",
    np.where(marks >= 40,"c",
    "fail"))

)
print(grades)
```

```python
count_fail = (grades == 'fail').sum()
count_fail
```

```python
z = np.where(grades == 'fail')
len(z)
```

```python
print(grades)
count_fail = len(np.where(grades == 'fail')[0])
count_fail
```

```python
np.sum(grades == 'fail')
```

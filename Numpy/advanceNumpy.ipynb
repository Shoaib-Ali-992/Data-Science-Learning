```python
import numpy as np



# Example true labels and predicted probabilities
y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0.1, 0.9, 0.8, 0.2, 0.7])

# Compute Binary Cross-Entropy using NumPy
def binary_cross_entropy(y_true, y_pred):
    bce = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return bce

bce_loss = binary_cross_entropy(y_true, y_pred)
print(f"Binary Cross-Entropy Loss (manual calculation): {bce_loss}")

# Compute Binary Cross-Entropy using Keras
```

```python
import numpy as np
import matplotlib.pyplot as plt

# True labels
y_true = np.array([0, 1, 1, 0, 1])

# x-axis (e.g., epochs)
x = np.arange(1, 6)

# Different predictions over time
y_preds = [
    np.array([0.2, 0.7, 0.6, 0.3, 0.8]),
    np.array([0.3, 0.8, 0.7, 0.2, 0.85]),
    np.array([0.4, 0.85, 0.75, 0.15, 0.9]),
    np.array([0.45, 0.9, 0.8, 0.1, 0.95]),
    np.array([0.5, 0.95, 0.85, 0.05, 0.97])
]

# Compute loss for each step
losses = []
for y_pred in y_preds:
    loss = -np.mean(
        y_true * np.log(y_pred) +
        (1 - y_true) * np.log(1 - y_pred)
    )
    losses.append(loss)

# Plot
plt.plot(x, losses, marker='o')
plt.xlabel("Epoch")
plt.ylabel("Binary Cross-Entropy Loss")
plt.title("Loss vs Epoch")
plt.show()
bce_loss = bce(y_true, y_pred)
print(f"Binary Cross_Entropy Loss (manual calculation): {bce_loss}")

```

# Iterating


```python
a1 = np.arange(10)
a1
```

```python
for i in a1:
    print(i)
```

```python
a2 = np.arange(12).reshape(3,4)
a2
```

```python
for i in a2:
    print(i)
```

```python
a3 = np.arange(27).reshape(3,3,3)
a3
```

```python
for i in a3:    # in 3d array loop run every bock indexing 
    print(i[2])
```

```python
for i in np.nditer(a3):  # nditer-----> firt convert in 1d array than loop run
    print(i)
```

```python
for i in np.ravel(a2):
    print(i)
```

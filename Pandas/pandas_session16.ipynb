### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### 
Importing Pandas

```python
import numpy as np
import pandas as pd
```

```python
# series from list
# string
country = ['Pakistan','India','USA','Nepal',]
pd.Series(country)
```

```python
# integers
runs = [12,44,44,33,56,89]
runs_ser = pd.Series(runs)
runs_ser
```

```python
# change data type
pd.Series(runs,dtype=np.int32)
```

```python
# custom index
marks = [39,56,78,89,90]
subjects = ['English','science','Urdu','math','data science']
# set all subjects in upper case
for idx in range(len(subjects)):
    subjects[idx] = subjects[idx].upper()
    
pd.Series(marks,index=subjects)
```

```python
# setting name
marks = pd.Series(marks, index=subjects,name='shoaib ke marks')
marks
```

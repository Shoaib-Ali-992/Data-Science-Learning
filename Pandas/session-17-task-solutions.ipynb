You need to make changes accordingly. Consider current name for each teams.

Be careful Gujrat Titans and Gujrat Lions are different teams. 

````python
data = pd.read_csv('IPL_Matches_2008_2022.csv')
change_name = {'Delhi Capitals':'Delhi Daredevils' ,
               'Punjab Kings':'Kings XI Punjab',
               'Rising Pune Supergiant':'Rising Pune Supergiants'
              }
data.replace(change_name.keys(),change_name.values(),inplace=True)
data.tail(10)
````

### `Q-6` Write a code which can display the bar chart of top 5 teams who have played maximum number of matches in the IPL.
>  Hint: Be careful the data is divided in 2 different cols(Team 1 and Team 2)



````python
data1 = data['Team2'].value_counts() + data['Team1'].value_counts()
data1.sort_values(ascending=False).head().plot(kind='bar')
````

````python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
# data = pd.read_csv("IPL_Matches_2008_2025.csv")  # replace with your CSV path

# Combine Team1 and Team2 into a single Series
all_teams = pd.concat([data['Team1'], data['Team2']])

# Count matches played by each team
team_counts = all_teams.value_counts()

# Get top 5 teams
top5_teams = team_counts.head(5)

# Plot bar chart
plt.figure(figsize=(10,6))
top5_teams.plot(kind='bar', color='skyblue')
plt.title("Top 5 IPL Teams by Number of Matches Played")
plt.xlabel("Team")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.show()
````

### `Q-8:` Team1 vs Team2 Dashbord
Create a function which will take two string(name of two teams) as input. Show win Loss record between them and player getting most player of the match award in matches between these two teams.
```
team1_vs_team2('Kolkata Knight Riders','Chennai Super Kings')
```

````python
def dasboard(t1,t2):
    m1 = (data.Team1 == t1) | (data.Team2==t1)
    m2 = (data.Team1 == t2) | (data.Team2==t2)
    df1 = data[m1 & m2]
    print(df1.WinningTeam.value_counts())
    print(df1.Player_of_Match.value_counts().head(1))
dasboard('Kolkata Knight Riders','Chennai Super Kings')

````

````python
ipl = pd.read_csv('https://docs.google.com/spreadsheets/d/e/2PACX-1vRy2DUdUbaKx_Co9F0FSnIlyS-8kp4aKv_I0-qzNeghiZHAI_hw94gKG22XTxNJHMFnFVKsO4xWOdIs/pub?gid=1655759976&single=true&output=csv')
# Replace old team name as new name before performing any tasks.
name_change = {
    'Delhi Daredevils':'Delhi Capitals',
    'Kings XI Punjab':'Punjab Kings',
    'Rising Pune Supergiants':'Rising Pune Supergiant'
}
ipl.replace(name_change,inplace=True)
# ✔ No-result matches removed
ipl = ipl[ipl['WinningTeam'].notna()]
# home team
home_team = ipl['Team1'].unique()
# away team
away_team = ipl['Team2'].unique()
# Matches Played = Team1 + Team2
matches_played = (ipl['Team1'].value_counts() + ipl['Team2'].value_counts())
winning_count = ipl['WinningTeam'].value_counts()
# Win %, Home Win %, Away Win %
win_percentage = (winning_count / matches_played)*100
# Home Win %
home_wins = ipl[ipl['Team1'] == ipl['WinningTeam']]['Team1'].value_counts()
home_matches = ipl['Team1'].value_counts()
home_persontage = (home_wins / home_matches)*100
 # Away Win %
away_wins = ipl[ipl['Team2'] == ipl['WinningTeam']]['Team2'].value_counts()
away_matches = ipl['Team2'].value_counts()
away_persontage = (away_wins / away_matches)*100
f = ipl[ipl['MatchNumber'] == 'Final']
winning_Tital = f['WinningTeam'].value_counts()
new_df = pd.DataFrame({
    'Teams': matches_played.index,
    'Matches_played':matches_played.values,
    'Win_percentage':win_percentage,
    'Home_percentage':home_persontage,
    'Away_percentage':away_persontage
})
new_df.index = range(1, len(new_df)+1)
new_df[['Win_percentage', 'Home_percentage', 'Away_percentage']] = new_df[['Win_percentage', 'Home_percentage', 'Away_percentage']].round(2)
new_df.sort_values(by='Win_percentage',ascending=False,inplace=True)
new_df
finals = ipl[ipl['MatchNumber'] == 'Final']
titles = finals['WinningTeam'].value_counts()
new_df['Titles_Won'] = new_df['Teams'].map(titles).fillna(0).astype(int)
new_df
````

````python
import pandas as pd

# Function to convert a player list string to a proper Python list
def getPlayers(player_str):
    return [p.strip(" '") for p in player_str.strip("[]").split(",")]

# Filter final matches
df_final = ipl[ipl["MatchNumber"] == "Final"]

# Apply getPlayers to both Team1Players and Team2Players
team1_lists = df_final['Team1Players'].apply(getPlayers)
team2_lists = df_final['Team2Players'].apply(getPlayers)

# Combine both columns into a single Series using concat and explode
all_players_series = pd.concat([team1_lists, team2_lists]).explode()

# Count player appearances
top_players = all_players_series.value_counts().head()

print(top_players)


````

````python
f_ipl = ipl[ipl['Season'] == '2009']
teams = f_ipl['Team1'].unique()
matches_played = (f_ipl['Team1'].value_counts() + f_ipl['Team2'].value_counts())
matches_win = f_ipl['WinningTeam'].value_counts()
no_result_per_team = (
    f_ipl[f_ipl['WonBy'] == 'NoResults']['Team1'].value_counts()
    .add(f_ipl[f_ipl['WonBy'] == 'NoResults']['Team2'].value_counts(), fill_value=0)
    .astype(int)
)
points = (
    matches_win.mul(2)
    .add(no_result_per_team, fill_value=0)
    .astype(int)
)
point_tables = {
    'Teams':matches_played.index,
    'Matches_Played': matches_played.values,
    'Wins':matches_win,
    'NO_Result':no_result_per_team,
    'Points':points
}
points_table = pd.DataFrame(point_tables)
points_table['NO_Result'].fillna(0,inplace=True)
# points_table.astype(int)
points_table.sort_values(by='Points',ascending=False,inplace=True)
# points_table['NO_Result'] = points_table['NO_Result'].map(no_result_per_team).fillna(0).astype(int)
# points_table.set_index('Teams',inplace=True)
points_table




````

````python
# Filter season
f_ipl = ipl[ipl['Season'] == '2009']

# Matches played
matches_played = (
    f_ipl['Team1'].value_counts()
    .add(f_ipl['Team2'].value_counts(), fill_value=0)
)

# Matches won
matches_win = f_ipl['WinningTeam'].value_counts()

# No Result per team
no_result_per_team = (
    f_ipl[f_ipl['WonBy'].str.strip() == 'NoResults']['Team1'].value_counts()
    .add(
        f_ipl[f_ipl['WonBy'].str.strip() == 'NoResults']['Team2'].value_counts(),
        fill_value=0
    )
)

# Points (2 per win + 1 per no result)
points = (
    matches_win.mul(2)
    .add(no_result_per_team, fill_value=0)
)

# ✅ Build points table properly
points_table = pd.DataFrame({
    'Matches_Played': matches_played,
    'Wins': matches_win,
    'No_Result': no_result_per_team,
    'Points': points
}).fillna(0).astype(int)

# Sort table
points_table.sort_values(by='Points', ascending=False, inplace=True)

points_table
````

````python
def matches_played(df, team):
    return df[(df.Team1 == team) | (df.Team2 == team)].shape[0]
def matches_won(df, team):
    return df[df.WinningTeam == team].shape[0]
def matches_no_result(df, team):
    return df[((df.Team1 == team) | (df.Team2 == team)) & (df.WinningTeam.isnull())].shape[0]

def point_table(season):
    df = ipl[ipl.Season == season]
    new_df = pd.DataFrame()
    new_df['Team Name'] = df.Team1.unique()
    new_df["Matches Played"] = new_df["Team Name"].apply(lambda x: matches_played(df, x))
    new_df["Matches Won"] = new_df["Team Name"].apply(lambda x: matches_won(df, x))
    new_df["No Result"] = new_df["Team Name"].apply(lambda x: matches_no_result(df, x))
    new_df["Points"] = new_df["Matches Won"]*2 + new_df["No Result"]
    new_df.sort_values("Points", ascending = False, inplace=True)
    new_df.set_index("Team Name", inplace=True)
    return new_df
point_table("2009")
````

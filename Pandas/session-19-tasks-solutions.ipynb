````python
# Make a copy
df = balls.copy()

# Create order-independent batting pairs
df["Batsman1"] = df[["batter", "non-striker"]].min(axis=1)
df["Batsman2"] = df[["batter", "non-striker"]].max(axis=1)
# Aggregate partnership stats
pair_df = df.groupby(["Batsman1", "Batsman2"]).agg(
    Runs=("total_run", "sum"),
    Balls=("ballnumber", "count"),   # wides counted as allowed
    Innings=("ID", "nunique")
).reset_index()


# Calculate Average & Strike Rate
pair_df["Avg"] = pair_df["Runs"] / pair_df["Innings"]
pair_df["StrikeRate"] = (pair_df["Runs"] / pair_df["Balls"]) * 100

# Final result
pair_df = pair_df.sort_values("Runs", ascending=False)
pair_df


# pair_df[["Batsman1", "Batsman2", "Runs", "Avg", "StrikeRate"]]
````

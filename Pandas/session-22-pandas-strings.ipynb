```python
# import titanic
df = pd.read_csv('titanic.csv')
df.info()
```

```python
df['Name']
```

```python
# Common Functions
# lower/upper/capitalize/title
df['Name'].str.capitalize()
df['Name'].str.upper()
df['Name'].str.lower()
df['Name'].str.title()
# extract large len of name
# len
df['Name'][df['Name'].str.len() == 82 ].values[0]
```

```python
df['Name'][df['Name'].str.len() == df['Name'].str.len().max()].values[0]
```

```python
# idxmax() returns ONLY the first occurrence of the maximum value
df.loc[df['Name'].str.len().idxmax(), 'Name']
```

```python
# strip()
'            malik         '.strip()
```

```python
df['Name'].str.strip()
```

```python
# split
# split -> get
df['last_name'] = df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title','first_name']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
df.head()
```

```python
df['title'].value_counts()
```

```python
# replace
df['title'] = df['title'].str.replace('Ms.','Miss.') 
df['title'] = df['title'].str.replace('Mlle.','Miss.') 
```

```python
df['title'].value_counts()
```

```python
# filtering
# startswith/endswith
df[df['first_name'].str.startswith('A')]
df[df['first_name'].str.endswith('A')]
```

```python
# isdigit/isalpha...
df[df['first_name'].str.isdigit()]
```

```python

df[df['first_name'].str.contains(r'\d')]
```

```python
df['first_name'] = df['first_name'].str.replace('Owen Harris', 'Owen Harris1')
```

```python
# You want rows where first_name contains ANY digit
df[df['first_name'].str.contains(r'\d')]
```

```python
# applying regex
# contains
# search john -> both case
df[df['first_name'].str.contains('john',case=False)]
# find lastnames with start and end char vowel
df[df['last_name'].str.contains('^[aeiouAEIOU].*[aeiouAEIOU]$',na=False)]
```

```python
# find lastnames with not start and end char vowel
df[df['last_name'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```
